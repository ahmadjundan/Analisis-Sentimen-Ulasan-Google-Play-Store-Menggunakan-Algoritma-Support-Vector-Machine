```python
#Import library
from google_play_scraper import app
import pandas as pd
import numpy as np
```

```python
#Pengambilan data ulasan
from google_play_scraper import Sort, reviews

result, continuation_token = reviews(
    'com.', #Ubah Menjadi Package name Aplikasi
    lang='id', #Ubah Menjadi Bahasa yang diinginkan
    country='id', #Ubah Menjadi Country Code
    sort=Sort.NEWEST, #Urutkan berdasarkan terbaru defaultnya MOST_RELEVANT
    count=20000, #Jumlah data yang diambil
    filter_score_with=None
)
```

```python
#Membuat DataFrame dan menampilkan data awal
df = pd.DataFrame(np.array(result),columns=['review'])

df = df.join(pd.DataFrame(df.pop('review').tolist()))

df.head()
```

```python
#Menampilkan data kolom rating, date-time, dan reviews
df[['at', 'content', 'score','appVersion']].head()
```

```python
#Mengurutkan data berdasarkan tanggal
df = df[['at', 'content', 'score', 'appVersion']]
df = df.sort_values(by='at', ascending=False)
```

```python
#Menyimpan data ke file csv
df.to_csv('scrapped.csv', index=False)
```
